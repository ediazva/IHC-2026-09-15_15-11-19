## 1. Verificación de una base de $\mathbb{R}^5$

Considere
$$
\begin{aligned}
v_1&=(1,1,0,1,2), & v_2&=(0,1,1,0,1),\\
v_3&=(1,0,1,1,0), & v_4&=(2,1,0,0,1),\\
v_5&=(1,2,1,1,1).
\end{aligned}
$$

¿Es $\mathcal B=\{v_1,v_2,v_3,v_4,v_5\}$ una base de $\mathbb R^5$?

Como hay cinco vectores en un espacio de dimensión $5$, basta verificar que sean linealmente independientes. Construimos la matriz
$$
A=\begin{pmatrix}|&|&|&|&|\\v_1&v_2&v_3&v_4&v_5\\|&|&|&|&|\end{pmatrix}.
$$
Entonces
$$
\mathcal B\text{ es base de }\mathbb R^5
\Longleftrightarrow \operatorname{rango}(A)=5 \Longleftrightarrow \det(A)\neq0.
$$

```python
import numpy as np

v1 = np.array([1, 1, 0, 1, 2])
v2 = np.array([0, 1, 1, 0, 1])
v3 = np.array([1, 0, 1, 1, 0])
v4 = np.array([2, 1, 0, 0, 1])
v5 = np.array([1, 2, 1, 1, 1])
# v5 = v1+v2

A = np.column_stack((v1, v2, v3, v4, v5))

print(A)
print("Rango =", np.linalg.matrix_rank(A))
print("Determinante =", np.linalg.det(A))
```

## 2. Polinomios como vectores

Un polinomio
$$
p(x)=a+bx+cx^2
$$
puede identificarse con su vector de coeficientes:
$$
a+bx+cx^2\longleftrightarrow(a,b,c).
$$

Esta correspondencia preserva suma y multiplicación por escalares. En general,
$$
a_0+a_1x+\cdots+a_nx^n
\longleftrightarrow
(a_0,a_1,\ldots,a_n).
$$

Por ello, los problemas de independencia lineal y bases en $P_n[x]$ pueden estudiarse mediante vectores de coeficientes.

### Ejemplo en $P_3[x]$

Considere
$$
p_1=1+x,\qquad p_2=x+x^2,\qquad
p_3=x^2+x^3,\qquad p_4=1-x+x^3.
$$

Respecto del orden $(1,x,x^2,x^3)$:
$$
p_1\leftrightarrow(1,1,0,0),\quad
p_2\leftrightarrow(0,1,1,0),
$$
$$
p_3\leftrightarrow(0,0,1,1),\quad
p_4\leftrightarrow(1,-1,0,1).
$$

Verifique si $\mathcal B=\{p_1,p_2,p_3,p_4\}$ es una base de $P_3[x]$.

```python
p1 = np.array([1, 1, 0, 0])
p2 = np.array([0, 1, 1, 0])
p3 = np.array([0, 0, 1, 1])
p4 = np.array([1, -1, 0, 1])

P = np.column_stack((p1, p2, p3, p4))

print(P)
print("Rango =", np.linalg.matrix_rank(P))
print("Determinante =", np.linalg.det(P))
```

Como $\dim(P_3[x])=4$, cuatro polinomios linealmente independientes forman una base.

$$
\boxed{\text{Podemos estudiar polinomios mediante sus vectores de coeficientes.}}
$$

## 3. Coordenadas respecto de distintas bases

Considere las bases de $\mathbb R^3$
$$
\mathcal B=\{(1,1,0),(0,1,1),(1,0,1)\}
$$
y
$$
\mathcal C=\{(1,0,0),(1,1,0),(1,1,1)\}.
$$

Generaremos un vector aleatorio $x\in\mathbb R^3$ y encontraremos sus coordenadas en ambas bases.

Si
$$
[x]_{\mathcal B}=
\begin{pmatrix}\alpha_1\\\alpha_2\\\alpha_3\end{pmatrix},
$$
entonces
$$
x=\alpha_1b_1+\alpha_2b_2+\alpha_3b_3.
$$
Si $B$ tiene como columnas los vectores de $\mathcal B$, debemos resolver
$$
B[x]_{\mathcal B}=x.
$$

```python
B = np.column_stack(([1, 1, 0], [0, 1, 1], [1, 0, 1]))
C = np.column_stack(([1, 0, 0], [1, 1, 0], [1, 1, 1]))

print("Rango de B =", np.linalg.matrix_rank(B))
print("Rango de C =", np.linalg.matrix_rank(C))

rng = np.random.default_rng(123)
x = rng.integers(-5, 6, size=3)
print("x =", x)

coord_B = np.linalg.solve(B, x)
coord_C = np.linalg.solve(C, x)

print("[x]_B =", coord_B)
print("[x]_C =", coord_C)
```

Aunque las coordenadas sean diferentes, ambas describen el mismo vector:
$$
x=B[x]_{\mathcal B}=C[x]_{\mathcal C}.
$$

```python
x_desde_B = B @ coord_B
x_desde_C = C @ coord_C

print("x =", x)
print("B[x]_B =", x_desde_B)
print("C[x]_C =", x_desde_C)

print("¿B[x]_B = x?", np.allclose(x_desde_B, x))
print("¿C[x]_C = x?", np.allclose(x_desde_C, x))
```
